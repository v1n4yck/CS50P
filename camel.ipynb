```python

#user enters in this format : firstName (basically camel format)
#we need to return : first_name (in snake format)



#lets say the user enters firstName
def main():
    variable_name= input("enter a name in camel case :")
    print(snake_case(split_name((variable_name))))


def split_name(m):
    x=""
    for char in m:
        if char.isupper():
            x= x + " " + char
        else:
            x=x + char
    return x
        


def snake_case(n):
    return n.replace(" ","_").lower()

main()
```
