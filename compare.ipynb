```python
#x a program that basically gets two int values from the user then compare them and states an obvious comparison like x is lesser than y or x is equal to y ,or x is greater thann y
# using CONDITIONALS : in which we basically ask the program a certain task based on some condition we give to the program only using if , else etc.

x = int(input("whats x ?"))
y = int(input("whats y ?"))

# if x > y:
#     print("x is greater than y")
# if x < y:
#     print("y is greater than x")
# if x == y:
#     print("x is equal to y")

# however there is a problem in this approach , we are making the interpretor hover in places it should, after it checks for the first statement that is x > y , and lets say it gets true from that , it would then be obvious that 
# there is no need to check the other two conditions cause there is no way in this case that we might get them true , here we will use elif which basically acts on the basis of the answer of the previous condition asks for ex:

if x > y:
    print("x is greater than y")
elif x < y:
    print("y is greater than x")
elif x == y:
    print("x is equal to y")

# now the change that occurred here is basically that we are not ALWAYS asking the same number of questions everytime now, now if the answer to first condition is true we are then directly stoppin the program without asking the other two questions for no reason like we were before. basically this change of using "elif" instead of "if" everywhere improved the control flow of the code a lot better. note that this is not a drastic change for a program so small


 
```
