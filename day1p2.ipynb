```python
#float

#approach 1
x = float(input("whats x ?"))
y = float(input("whats y "))
z = round(x / y,2) #here we are telling it to round the float number upto two decimal places
print(z)

# round method, used to round off to the nearest
```

```python
#another approach, cryptic one
x = float(input("whats x ?"))
y = float(input("whats y "))
z = x / y
print(f"{z:.2f}") # :.2f this whole thing basically tells python to round up the float upto two decimal places
```

```python
#create a function that says hello to the user
#def keyword : used for defining new functions 
def hello(to= "world"):#empty parenthesis will mean the function wont take any arguments             #we can also give the "to" parameter a default value for incase the hello function isnt given any input
    print(f"hello, {to}")#here to is given a default value , which will act only if the hello function is NOT given any input 
hello()
name = input("whats your name ?")
hello(name)
```

```python
#better approach to write a code 
def main():
    name= input("whats your name ?")
    hello(name)

def hello(x):
    print(f"hello, {x}")

main()

#why this is the better approach : there is a thing called scope, if we call hello function in lets say line 8 but the function is defined at line 10, it will throw an error obviously. 
#so this approach fixes that complication and basically sums everything up into a main function which basically holds the key to all other functions that we need right at the start of the code,
#  main is basically the main part of our code
```
