```python
# making a program  that assigns a grade to the percentage of the score that the user scored in exams 
score = int(input("enter your score :"))

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
elif score >= 60:
    print("Grade D")
elif score >= 50:
    print("Grade E")
else :print("Grade F")
```
