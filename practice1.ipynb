```python
#type conversion : 1) implicit 2) explicit
#implicit : type of conversion which python does on its own without being told to, for example
print(2 + 1.5)
# here 2 is a integer and 1.5 is a float , yet we didnt have to explicitly tell python to convert the integer 2 into a float, it did it on its own.

#explicit
num_1 = int(input("enter the first number"))
num_2 = int(input("enter the second number"))
result = num_1+num_2
print(result)

#here it didnt do it on its own , we had to tell it to do the type conversion of input from string to integer
```
